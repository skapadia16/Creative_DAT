```python
def convertRange(x):
    temp= str(x).split('-')
    if len(temp) == 2:
     return (float(temp[0]) + float(temp[1])) / 2
    try:
        return float(x)
    except:
        return None
```

```python
data['total_sqft'] = data['total_sqft'].apply(convertRange)
data['total_sqft'] = pd.to_numeric(data['total_sqft'], errors='coerce')
data.head()
```

```python
def remove_outlier(df):
    df_output = pd.DataFrame()
    for kry,subdf in df.groupby('location'): ##groupby lagavi aetle key aape and aeno subdata aape subdf aetle subdata
        
        m = np.mean(subdf.price_per_sqft)

        st = np.std(subdf.price_per_sqft)

        gen_df = subdf[(subdf.price_per_sqft> (m-st)) & (subdf.price_per_sqft <= (m+st))]
        df_output = pd.concat([df_output,gen_df],ignore_index=True)
    return df_output
data = remove_outlier(data)
data.describe()
```

```python
def bhk_outlier (df):
    exclude_indices = np.array([])
    for location,location_df in df.groupby("location"):
        bhk_stats = {}
        for bhk,bhk_df in location_df.groupby('bhk'):
            bhk_stats[bhk] = {
                "mean": np.mean(bhk_df.price_per_sqft),
                "std": np.std(bhk_df.price_per_sqft),
                "count": bhk_df.shape[0]}
        for bhk, bhk_df in location_df.groupby('bhk'):
            stats = bhk_stats.get(bhk-1)
            if stats and stats['count']>5:
                exclude_indices = np.append(exclude_indices,bhk_df[bhk_df.price_per_sqft<(stats['mean'])].index.values)
    return df.drop(exclude_indices,axis='index')
```

```python
x = data.drop(columns=["price"])
y = data["price"]
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression,Lasso,Ridge
from sklearn.preprocessing import OneHotEncoder,StandardScaler
from sklearn.compose import make_column_transformer
from sklearn.pipeline import make_pipeline
from sklearn.metrics import r2_score
```

```python
x_train, x_test, y_train , y_test = train_test_split(x,y,test_size=0.2,random_state=23)
```

```python
column_trans = make_column_transformer((OneHotEncoder(sparse_output=False),['location']),
                                       remainder = 'passthrough')
```

```python
scaler = StandardScaler()
```

```python
lr = LinearRegression()
```

```python
pipe = make_pipeline(column_trans,scaler,lr)
```

```python
pipe.fit(x_train,y_train)
```

```python
y_pred = pipe.predict(x_test)
```

```python
r2_score(y_test,y_pred)
```

```python
# Load model
model = pickle.load(open('ridgemodel.pkl', 'rb'))

# Get only trained locations
locations = model.named_steps['columntransformer'] \
                 .named_transformers_['onehotencoder'] \
                 .categories_[0]

locations = sorted(locations)

# Widgets
import ipywidgets as widgets
from IPython.display import display
import pandas as pd

area = widgets.FloatText(description="Area")
bed = widgets.IntText(description="Bedrooms")
bath = widgets.IntText(description="Bathrooms")

location = widgets.Dropdown(
    options=locations,
    description="Location"
)

output = widgets.Output()
button = widgets.Button(description="Predict")

def predict(b):
    with output:
        output.clear_output()
        try:
            input_data = pd.DataFrame({
                'total_sqft': [area.value],
                'bhk': [bed.value],
                'bath': [bath.value],
                'location': [location.value]
            })

            result = model.predict(input_data)
            price = result[0]

            if price < 0:
                print("⚠️ Invalid input / unrealistic house details")
            else:
                print("Price:", round(price, 2),"Lacs")

        except Exception as e:
            print("Error:", e)

button.on_click(predict)

display(area, bed, bath, location, button, output)
```
